```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
spark.sql(f"""CREATE OR REPLACE TABLE {catalog}.{schema}.customer_360_new AS
SELECT country,
       COUNT(*) AS number_of_customer,
       COUNT(CASE WHEN gender='female' THEN 1 END) AS female_customers,
       COUNT(CASE WHEN gender='male' THEN 1 END) AS male_customers
FROM samples.bakehouse.sales_customers
GROUP BY country""")
```
